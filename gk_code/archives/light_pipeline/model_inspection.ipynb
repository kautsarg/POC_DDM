```python
import tensorflow as tf
exp_folder = Path(config.MULTI_EXP_FOLDERS)
exp_paths = sorted([p for p in exp_folder.iterdir() if p.is_dir() and p.name not in ['.DS_Store', 'model_interpretation']])
exp_path = exp_paths[0]
model_dir = exp_path / "model_interpretation"
filter_key = None

data_dict = prepare_dataset(exp_path, filter_key)
expected_seq_len = data_dict['X_full'].shape[1]

models = load_saved_models(model_dir, str(filter_key), expected_seq_len)
```

```python
for name, model in models.items():
    output_img_path = model_dir / f"{name}_architecture.png"
    
    print("=" * 60)
    print(f" ARCHITECTURE SUMMARY FOR MODEL: {name.upper()}")
    print("=" * 60)
    
    # Prints layers, output shapes, and trainable vs non-trainable parameters
    model.summary()
    print("\n" * 2)
    
    tf.keras.utils.plot_model(
        model,
        # to_file=str(output_img_path),
        show_shapes=True,        # Highly recommended: displays input/output dimensions per layer
        show_layer_names=True,   # Displays layer types (Conv1D, GRU, Dense, Attention, etc.)
        rankdir='TB',            # 'TB' = Top-to-Bottom flow, 'LR' = Left-to-Right flow
        dpi=96                   # Adjust resolution sharpness
    )
    # print(f"    [✓] Saved image to: {output_img_path}")
```
